## 7. Method behind the prototype

| Discipline | How it appears here |
|---|---|
| **Business + data understanding** | Define 20/week and 160 total, then inspect the existing recruitment and attendance funnel. |
| **CRISP-DM** | Understand → inspect data → prepare → model/forecast → evaluate → operate. |
| **Iterative SDLC** | Implement one bounded feature, test it, compare with the baseline, and expand. |
| **Data pyramid** | Synthetic source rows → validated records → weekly/channel/student views → management decision. |
| **OOP** | `CampaignMonitor`, `DecisionEngine`, `StudentProfile`, and `StudentSupportEngine` keep responsibilities separate and testable. |

AI can help research partners, draft outreach variants, adapt seminar content, summarize feedback, and detect patterns. Ordinary rules and workflow automation handle deterministic steps; AI is used only where it adds value.
